```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('Churn_Unclean_Project.xlsx')
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.replace(['N/A','NULL','' ,' '],np.nan,inplace=True)
```

```python
df.isnull().sum()
```

```python
df=df.apply(lambda x:x.str.strip() if x.dtype=="object" else x)
```

```python
df["Subscription_Type"].value_counts()
```

```python
df["Customer_Name"]=df["Customer_Name"].str.title()
df["State"]=df["State"].str.title()
df["City"]=df["City"].str.title()
df["Subscription_Type"]=df["Subscription_Type"].str.title()
df["Contract_Type"]=df["Contract_Type"].str.title()
```

```python
df["Subscription_Type"].value_counts()
```

```python
# df["Subscription_Type"] = df["Subscription_Type"].str.strip()
df["Subscription_Type"] = df["Subscription_Type"].str.title()
```

```python
df["Subscription_Type"] = df["Subscription_Type"].str.strip().str.lower()
df["Subscription_Type"] = df["Subscription_Type"].replace({
    "premium": "Premium",
    "basic": "Basic",
    "standard": "Standard"
})
```

```python
df["Subscription_Type"].value_counts()
```

```python
df["Churn"].value_counts()
```

```python
df["Churn"]=df["Churn"].str.title()
```

```python
df["Gender"]=df["Gender"].str.title()
```

```python
df["Gender"].value_counts()
```

```python
cols=[
"Age",
"Tenure_Months",
"Monthly_Charges",
"Total_Charges"
]
for c in cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")
```

```python
df.describe()
```

```python
df=df[(df["Age"]>=18)&(df["Age"]<=100)]
```

```python
df=df[df["Monthly_Charges"]>=0]
df=df[df["Total_Charges"]>=0]
```

```python
df.shape
```

```python
df["Last_Interaction_Date"]=pd.to_datetime(
    df["Last_Interaction_Date"],
    errors="coerce"
)
```

```python
# Fill Missing Values
df["Tech_Support"]=df["Tech_Support"].fillna("No")
df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")
df["Monthly_Charges"]=df["Monthly_Charges"].fillna(df['Monthly_Charges'].mean())
df["Tenure_Months"]=df["Tenure_Months"].fillna(df['Tenure_Months'].mean())
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df["Customer_Value"]=df["Monthly_Charges"]*df["Tenure_Months"]
```

```python
df.head()
```

```python
df["Monthly_Revenue"]=df["Monthly_Charges"]
```

```python
df["Tenure_Group"]=pd.cut(
df["Tenure_Months"],
bins=[0,12,24,48,72],
labels=["0-12","13-24","25-48","49-72"]
)
```

```python
df["Senior_Flag"]=np.where(df["Age"]>=60,"Senior","Adult")
```

```python
df["Churn_Flag"]=df["Churn"].map({
"Yes":1,
"No":0
})
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df["Last_Interaction_Date"]=df["Last_Interaction_Date"].fillna(df["Last_Interaction_Date"].median())
```

```python
df['Tenure_Months'] = df['Tenure_Months'].astype(int)
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df.to_csv("Clean_Churn_Data.csv",index=False)
```
